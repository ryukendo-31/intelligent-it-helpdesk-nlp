```python
import pandas as pd
import numpy as np

DATA_PATH = "../data/dataset-tickets-multi-lang3-4k.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
print("First 5 rows:")
display(df.head())

print("\nRandom 5 rows:")
display(df.sample(5, random_state=42))
```

```python
print("Queue Distribution:")
print(df["queue"].value_counts())

print("\nPriority Distribution:")
print(df["priority"].value_counts())

print("\nTicket Type Distribution:")
print(df["type"].value_counts())

print("\nLanguage Distribution:")
print(df["language"].value_counts())
```

```python
print("Number of unique queues:", df["queue"].nunique())
print("Number of unique priorities:", df["priority"].nunique())
print("Number of unique types:", df["type"].nunique())
print("Number of unique languages:", df["language"].nunique())

print("\nUnique queues:")
print(df["queue"].unique())

print("\nUnique priorities:")
print(df["priority"].unique())

print("\nUnique types:")
print(df["type"].unique())
```

```python
for i in range(3):
    print("=" * 80)
    print("SUBJECT:")
    print(df.iloc[i]["subject"])

    print("\nBODY:")
    print(df.iloc[i]["body"])

    print("\nQUEUE:", df.iloc[i]["queue"])
    print("PRIORITY:", df.iloc[i]["priority"])
    print("TYPE:", df.iloc[i]["type"])

    print("\nAGENT ANSWER:")
    print(df.iloc[i]["answer"])
```

```python
df["subject_length"] = df["subject"].fillna("").str.len()
df["body_length"] = df["body"].fillna("").str.len()
df["answer_length"] = df["answer"].fillna("").str.len()

print(df[["subject_length", "body_length", "answer_length"]].describe())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))
plt.hist(df["body_length"], bins=50)
plt.xlabel("Body Length")
plt.ylabel("Number of Tickets")
plt.title("Distribution of Ticket Body Length")
plt.show()
```
